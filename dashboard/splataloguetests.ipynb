```python
frequency_ranges = [(680.35, 682.34), (681.29, 683.15), (688.19, 690.06)]  # Example range pairs

all_lines = []
#loop through the freq pairs
for start_freq, end_freq in frequency_ranges:
    #Splat query
    lines = Splatalogue.query_lines(start_freq*u.GHz, end_freq*u.GHz, only_NRAO_recommended=True)
    lines_df = lines.to_pandas()
  
    #append df to list
    all_lines.append(lines_df)

#concatenate all df into big df
combined_df = pd.concat(all_lines, ignore_index=True)

combined_df['chemical_name'] = combined_df['chemical_name'].str.lower()
combined_df
```

```python
def q25(x):
    return x.quantile(0.25)

def q50(x):
    return x.quantile(0.5)
    
def q75(x):
    return x.quantile(0.75)
```

```python
x=pd.DataFrame(combined_df.groupby('chemical_name').agg(count=('chemical_name', 'count'), 
    lower_state_energy_K_25=('lower_state_energy_K', q25), 
    lower_state_energy_K_50=('lower_state_energy_K', q50), 
    lower_state_energy_K_75=('lower_state_energy_K', q75)).sort_values(by='count',ascending = False).reset_index())#.head(20).round(2)
x 
```
